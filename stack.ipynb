```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None
```

```python
class Stack:
    def __init__(self):
        self.top = None
        
    def isEmpty(self):
        return self.top == None
    
    def push(self,value):
        new_node = Node(value)
        new_node.next = self.top
        self.top = new_node
    def peak(self):
        if (self.isEmpty()):
            return "stack is empty"
        else:
            return self.top.data
    def pop(self):
        if (self.isEmpty()):
            return "stack is empty"
        else:
            self.top = self.top.next
        
    def traverse(self):
        temp = self.top
        while temp!=None:
            print(temp.data)
            temp = temp.next
```

```python
s = Stack()
s.isEmpty()
```

```python
s.push(2)
s.push(6)
s.push(4)
s.push(4)
s.push(3)
```

```python
s.traverse()
```

```python
s.pop()
s.peak()
```
